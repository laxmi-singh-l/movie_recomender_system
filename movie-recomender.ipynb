```python

import nltk
nltk.download('stopwords')

import re
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

def clean_and_stem(text):
    if not isinstance(text, str):
        return ""

    text = text.lower()
    text = re.sub(r'[^a-zA-Z\s]', '', text)  # remove punctuation & numbers

    words = text.split()
    words = [w for w in words if w not in stop_words]
    words = [ps.stem(w) for w in words]

    return " ".join(words)
```

# For repeating words

```python
import nltk
from nltk.stem.porter import PorterStemmer

ps = PorterStemmer()
```

```python
nmo['tags'] = nmo['tags'].apply(clean_and_stem)
```

```python
def stem(text):
  y = []

  for i in text.split():
    y.append(ps.stem(i))

  return " ".join(y)
```

In this case for similarity we are going to calculate `cosine distance` rather then point distence `(y^2 - Y^2)/(x^2 - X^2)`

using `coisne` similarity

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vectors)
```

```python
def recommend(movie):
  movie_index = nmo[nmo['title'] == movie].index[0]
  distances = similarity[movie_index]
  movies_list = sorted(list(enumerate(distances)),reverse= True, key = lambda x:x[1])[1:6]

  for i in movies_list:
    print(nmo.iloc[i[0]].title)
```

```python
recommend('Batman Begins')
```
